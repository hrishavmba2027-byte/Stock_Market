# GPU Stock Prediction Pipeline

This notebook is shape-safe by construction:

- Every stock is normalized to the same numeric feature schema.
- Sequences always have shape `(samples, seq_len, features)`.
- Dense input size is inferred from `X.shape[1] * X.shape[2]`.
- LSTM and Transformer feature dimensions are inferred from `X.shape[2]`.
- Each model validates batch shape before training and before every forward pass.
- Stock-level failures are logged and skipped so valid stocks continue through the pipeline.

```python

# ============================================================================
# Stock Normalization and Common Feature Alignment
# ============================================================================

PRICE_PREFIXES = {
    "open_": "Open",
    "high_": "High",
    "low_": "Low",
    "close_": "Close",
    "volume_": "Volume",
}
EXACT_PRICE_COLUMNS = {
    "open": "Open",
    "high": "High",
    "low": "Low",
    "close": "Close",
    "volume": "Volume",
}

def canonical_column_name(column):
    name = str(column).strip()
    lower = name.lower()
    if lower in EXACT_PRICE_COLUMNS:
        return EXACT_PRICE_COLUMNS[lower]
    for prefix, canonical in PRICE_PREFIXES.items():
        if lower.startswith(prefix):
            return canonical
    return name

def detect_date_column(df):
    date_candidates = [col for col in df.columns if "date" in str(col).lower() or "time" in str(col).lower()]
    return date_candidates[0] if date_candidates else None

def normalize_stock_sheet(df, symbol):
    df = df.copy()
    df.columns = [canonical_column_name(col) for col in df.columns]
    df = df.loc[:, ~pd.Index(df.columns).duplicated()].copy()

    date_col = detect_date_column(df)
    if date_col is not None:
        parsed_dates = pd.to_datetime(df[date_col], errors="coerce")
        if parsed_dates.notna().any():
            df[date_col] = parsed_dates
            df = df.sort_values(date_col).reset_index(drop=True)

    drop_cols = [col for col in df.columns if "date_str" in str(col).lower() or "date_string" in str(col).lower()]
    if drop_cols:
        df = df.drop(columns=drop_cols, errors="ignore")

    for col in df.columns:
        if col == date_col:
            continue
        converted = pd.to_numeric(df[col], errors="coerce")
        if converted.notna().sum() > 0:
            df[col] = converted

    if TARGET_COL not in df.columns:
        raise ValueError(f"Missing required target column '{TARGET_COL}' after normalization")

    df[TARGET_COL] = pd.to_numeric(df[TARGET_COL], errors="coerce")
    df = df.dropna(subset=[TARGET_COL]).reset_index(drop=True)

    numeric_features = [
        col for col in df.select_dtypes(include=[np.number]).columns
        if col != TARGET_COL
    ]
    if not numeric_features:
        raise ValueError("No numeric feature columns found")

    return df, numeric_features

stock_frames = {}
stock_feature_sets = {}
skipped_stocks = {}

for sheet_name in tqdm(xls.sheet_names, desc="Normalizing sheets"):
    try:
        raw_df = pd.read_excel(xls, sheet_name=sheet_name)
        engineered_df = compute_indicators(raw_df, detect_date_column(raw_df))
        normalized_df, numeric_features = normalize_stock_sheet(engineered_df, sheet_name)
        stock_frames[sheet_name] = normalized_df
        stock_feature_sets[sheet_name] = set(numeric_features)
        print(f"{sheet_name}: rows={len(normalized_df)}, numeric_features={len(numeric_features)}")
    except Exception as exc:
        skipped_stocks[sheet_name] = str(exc)
        print(f"Skipping {sheet_name}: {exc}")

if not stock_frames:
    raise RuntimeError("No valid stock sheets were loaded")

common_features = sorted(set.intersection(*stock_feature_sets.values()))
if not common_features:
    raise RuntimeError("No common numeric features found across valid stocks")

if USE_INTRADAY_OPEN:
    if "Open" not in common_features:
        raise RuntimeError("Open must be present in common_features for intraday prediction")
    if TARGET_COL in common_features:
        raise RuntimeError("Close must never appear in feature columns for intraday prediction")

print(f"\nValid stocks: {len(stock_frames)}")
print(f"Skipped stocks: {len(skipped_stocks)}")
print(f"Common feature count: {len(common_features)}")
print(f"Common features: {common_features}")
```

```python

# ============================================================================
# Sequence Builder with Leakage Guards
# ============================================================================

def clean_aligned_stock_frame(df, feature_columns, target_column):
    aligned = df.reindex(columns=feature_columns + [target_column], fill_value=0).copy()
    aligned = aligned.apply(pd.to_numeric, errors="coerce")
    aligned = aligned.replace([np.inf, -np.inf], np.nan)
    aligned = aligned.ffill().fillna(0)
    return aligned

def validate_feature_contract(feature_columns):
    if TARGET_COL in feature_columns:
        raise ValueError("Close is the target and must not be present in feature columns")
    if USE_INTRADAY_OPEN and "Open" not in feature_columns:
        raise ValueError("Open feature is required when USE_INTRADAY_OPEN is True")

def validate_split_bounds(row_count, train_end_idx, val_end_idx, seq_len):
    if not (seq_len < train_end_idx < val_end_idx < row_count):
        raise ValueError(
            f"Invalid chronological split: row_count={row_count}, "
            f"seq_len={seq_len}, train_end={train_end_idx}, val_end={val_end_idx}"
        )

def validate_no_leakage_state(history_indices, target_index, feature_columns):
    validate_feature_contract(feature_columns)
    history_indices = np.asarray(history_indices, dtype=np.int64)
    if history_indices.size == 0:
        raise ValueError("History cannot be empty for a prediction sample")
    if int(history_indices.max()) >= int(target_index):
        raise ValueError(
            f"Data leakage detected: history max index {int(history_indices.max())} "
            f"must be < target index {int(target_index)}"
        )

def validate_intraday_open_injection(before_seq, after_seq, open_idx):
    before_seq = np.asarray(before_seq)
    after_seq = np.asarray(after_seq)
    if before_seq.shape != after_seq.shape:
        raise ValueError(f"Open injection shape mismatch: {before_seq.shape} vs {after_seq.shape}")
    changed = np.argwhere(~np.isclose(before_seq, after_seq, rtol=0.0, atol=1e-7, equal_nan=True))
    if changed.size == 0:
        return
    allowed = (changed[:, 0] == before_seq.shape[0] - 1) & (changed[:, 1] == open_idx)
    if not np.all(allowed):
        raise ValueError(f"Open injection changed non-Open cells: {changed.tolist()}")

def snapshot_scaler(scaler):
    snapshot = {}
    for attr in ("data_min_", "data_max_", "scale_", "min_"):
        if hasattr(scaler, attr):
            snapshot[attr] = getattr(scaler, attr).copy()
    snapshot["n_samples_seen_"] = int(getattr(scaler, "n_samples_seen_", -1))
    return snapshot

def assert_scaler_unchanged(name, scaler, snapshot):
    for attr, expected in snapshot.items():
        actual = int(getattr(scaler, attr, -1)) if attr == "n_samples_seen_" else getattr(scaler, attr)
        if attr == "n_samples_seen_":
            if actual != expected:
                raise ValueError(f"{name} changed during walk-forward: {attr} {actual} != {expected}")
        elif not np.allclose(actual, expected, rtol=0.0, atol=1e-12, equal_nan=True):
            raise ValueError(f"{name} changed during walk-forward: {attr}")

def validate_scaler_fit_scope(scaler, expected_count, name):
    seen = int(getattr(scaler, "n_samples_seen_", -1))
    if seen != int(expected_count):
        raise ValueError(f"{name} fit scope mismatch: expected {expected_count} rows, saw {seen}")

def split_bounds_for_rows(row_count, seq_len):
    train_end_idx = int(TRAIN_RATIO * row_count)
    val_end_idx = int((TRAIN_RATIO + VAL_RATIO) * row_count)
    validate_split_bounds(row_count, train_end_idx, val_end_idx, seq_len)
    return train_end_idx, val_end_idx

def build_sequences(
    aligned_df,
    feature_columns,
    target_column,
    seq_len,
    feature_scaler,
    target_scaler,
    use_intraday_open=USE_INTRADAY_OPEN,
):
    validate_feature_contract(feature_columns)
    numeric_df = aligned_df[feature_columns + [target_column]].select_dtypes(include=[np.number])

    expected_columns = len(feature_columns) + 1
    if numeric_df.shape[1] != expected_columns:
        raise ValueError(f"Expected {expected_columns} numeric columns, got {numeric_df.shape[1]}")

    if len(numeric_df) <= seq_len:
        empty_X = np.empty((0, seq_len, len(feature_columns)), dtype=np.float32)
        empty_y = np.empty((0,), dtype=np.float32)
        empty_idx = np.empty((0,), dtype=np.int64)
        return empty_X, empty_y, empty_idx

    feature_frame = numeric_df[feature_columns]
    scaled_values = feature_scaler.transform(feature_frame).astype(np.float32)
    target_values = target_scaler.transform(numeric_df[[target_column]]).astype(np.float32).reshape(-1)
    open_idx = feature_columns.index("Open") if use_intraday_open else None

    X_sequences = []
    y_values = []
    target_indices = []

    for end_idx in range(seq_len, len(numeric_df)):
        history_indices = np.arange(end_idx - seq_len, end_idx, dtype=np.int64)
        validate_no_leakage_state(history_indices, end_idx, feature_columns)

        past_seq = scaled_values[end_idx - seq_len:end_idx].copy()

        if use_intraday_open:
            before_seq = past_seq.copy()
            open_t_scaled = scaled_values[end_idx][open_idx]
            past_seq[-1][open_idx] = open_t_scaled
            validate_intraday_open_injection(before_seq, past_seq, open_idx)

        X_sequences.append(past_seq)
        y_values.append(target_values[end_idx])
        target_indices.append(end_idx)

    X = np.stack(X_sequences).astype(np.float32)
    y = np.asarray(y_values, dtype=np.float32)
    target_indices = np.asarray(target_indices, dtype=np.int64)

    if X.ndim != 3:
        raise ValueError(f"Sequence builder returned invalid X shape: {X.shape}")
    if X.shape[1] != seq_len or X.shape[2] != len(feature_columns):
        raise ValueError(
            f"Sequence shape mismatch: expected (*, {seq_len}, {len(feature_columns)}), got {X.shape}"
        )
    if len(X) != len(y) or len(X) != len(target_indices):
        raise ValueError("X/y/index length mismatch")

    return X, y, target_indices

def split_stock_sequences(X, y, target_indices, symbol, train_end_idx, val_end_idx):
    target_indices = np.asarray(target_indices, dtype=np.int64)
    masks = {
        "train": target_indices < train_end_idx,
        "val": (target_indices >= train_end_idx) & (target_indices < val_end_idx),
        "test": target_indices >= val_end_idx,
    }

    splits = {}
    for split_name, mask in masks.items():
        split_count = int(mask.sum())
        if split_count == 0:
            raise ValueError(f"No {split_name} samples after chronological split")
        splits[split_name] = (
            X[mask],
            y[mask],
            np.full(split_count, symbol, dtype=object),
            target_indices[mask],
        )
    return splits

split_parts = {
    "train": {"X": [], "y": [], "symbols": [], "target_indices": []},
    "val": {"X": [], "y": [], "symbols": [], "target_indices": []},
    "test": {"X": [], "y": [], "symbols": [], "target_indices": []},
}
stock_feature_scalers = {}
stock_target_scalers = {}
stock_walk_forward_artifacts = {}
sequence_failures = {}

for symbol, df in tqdm(stock_frames.items(), desc="Building stock sequences"):
    try:
        aligned = clean_aligned_stock_frame(df, common_features, TARGET_COL)
        train_end_idx, val_end_idx = split_bounds_for_rows(len(aligned), SEQ_LEN)

        feature_scaler = MinMaxScaler()
        target_scaler = MinMaxScaler()
        feature_scaler.fit(aligned.iloc[:train_end_idx][common_features])
        target_scaler.fit(aligned.iloc[:train_end_idx][[TARGET_COL]])
        validate_scaler_fit_scope(feature_scaler, train_end_idx, f"{symbol} feature_scaler")
        validate_scaler_fit_scope(target_scaler, train_end_idx, f"{symbol} target_scaler")

        stock_feature_scalers[symbol] = feature_scaler
        stock_target_scalers[symbol] = target_scaler

        X_stock, y_stock, target_indices = build_sequences(
            aligned,
            common_features,
            TARGET_COL,
            SEQ_LEN,
            feature_scaler=feature_scaler,
            target_scaler=target_scaler,
        )
        if len(X_stock) == 0:
            raise ValueError(f"Not enough rows to build sequences with SEQ_LEN={SEQ_LEN}")

        stock_input_size = X_stock.shape[1] * X_stock.shape[2]
        print(
            f"{symbol} -> rows={len(aligned)}, split=0:{train_end_idx}:{val_end_idx}:{len(aligned)}, "
            f"X shape={X_stock.shape} -> input_size={stock_input_size}"
        )

        stock_split = split_stock_sequences(X_stock, y_stock, target_indices, symbol, train_end_idx, val_end_idx)
        for split_name, (X_part, y_part, symbols_part, indices_part) in stock_split.items():
            split_parts[split_name]["X"].append(X_part)
            split_parts[split_name]["y"].append(y_part)
            split_parts[split_name]["symbols"].append(symbols_part)
            split_parts[split_name]["target_indices"].append(indices_part)

        stock_walk_forward_artifacts[symbol] = {
            "source_frame": df.reset_index(drop=True).copy(),
            "date_col": detect_date_column(df),
            "feature_scaler": feature_scaler,
            "target_scaler": target_scaler,
            "train_end_idx": train_end_idx,
            "val_end_idx": val_end_idx,
            "test_target_indices": stock_split["test"][3],
        }
    except Exception as exc:
        sequence_failures[symbol] = str(exc)
        print(f"Skipping {symbol}: {exc}")

def concat_split(split_name):
    if not split_parts[split_name]["X"]:
        raise RuntimeError(f"No {split_name} sequences were created")
    X = np.concatenate(split_parts[split_name]["X"], axis=0).astype(np.float32)
    y = np.concatenate(split_parts[split_name]["y"], axis=0).astype(np.float32)
    symbols = np.concatenate(split_parts[split_name]["symbols"], axis=0)
    target_indices = np.concatenate(split_parts[split_name]["target_indices"], axis=0).astype(np.int64)
    if X.ndim != 3:
        raise ValueError(f"{split_name} X must be 3D, got {X.shape}")
    if X.shape[1] != SEQ_LEN or X.shape[2] != len(common_features):
        raise ValueError(f"{split_name} X has inconsistent shape: {X.shape}")
    return X, y, symbols, target_indices

X_train, y_train, train_symbols, train_target_indices = concat_split("train")
X_val, y_val, val_symbols, val_target_indices = concat_split("val")
X_test, y_test, test_symbols, test_target_indices = concat_split("test")

actual_seq_len = X_train.shape[1]
actual_feature_count = X_train.shape[2]
actual_input_size = X_train.shape[1] * X_train.shape[2]

print("\nGlobal tensor shapes")
print(f"Train X: {X_train.shape}, y: {y_train.shape}")
print(f"Val   X: {X_val.shape}, y: {y_val.shape}")
print(f"Test  X: {X_test.shape}, y: {y_test.shape}")
print(f"actual_input_size = X.shape[1] * X.shape[2] = {actual_seq_len} * {actual_feature_count} = {actual_input_size}")
print(f"Sequence failures: {len(sequence_failures)}")
```

```python

# ============================================================================
# Dataset and DataLoaders
# ============================================================================

class StockDataset(Dataset):
    def __init__(self, X, y):
        if not isinstance(X, np.ndarray):
            X = np.asarray(X, dtype=np.float32)
        if not isinstance(y, np.ndarray):
            y = np.asarray(y, dtype=np.float32)

        if X.ndim != 3:
            raise ValueError(f"X must be 3D (samples, seq_len, features), got {X.shape}")
        if y.ndim != 1:
            y = y.reshape(-1)
        if len(X) != len(y):
            raise ValueError(f"X/y sample mismatch: {len(X)} vs {len(y)}")
        if X.shape[1] != actual_seq_len or X.shape[2] != actual_feature_count:
            raise ValueError(
                f"Dataset shape mismatch: expected (*, {actual_seq_len}, {actual_feature_count}), got {X.shape}"
            )

        self.X = torch.as_tensor(X, dtype=torch.float32).contiguous()
        self.y = torch.as_tensor(y, dtype=torch.float32).view(-1, 1).contiguous()

    def __len__(self):
        return self.X.shape[0]

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

def make_loader(dataset, shuffle=False):
    if shuffle:
        raise ValueError("Shuffling is disabled for strict chronological time-series training")
    loader_kwargs = {
        "batch_size": BATCH_SIZE,
        "shuffle": False,
        "num_workers": NUM_WORKERS,
        "pin_memory": PIN_MEMORY,
        "drop_last": False,
    }
    if NUM_WORKERS > 0:
        loader_kwargs.update({"prefetch_factor": 2, "persistent_workers": True})
    return DataLoader(dataset, **loader_kwargs)

train_dataset = StockDataset(X_train, y_train)
val_dataset = StockDataset(X_val, y_val)
test_dataset = StockDataset(X_test, y_test)

train_loader = make_loader(train_dataset, shuffle=False)
val_loader = make_loader(val_dataset, shuffle=False)
test_loader = make_loader(test_dataset, shuffle=False)

for split_name, loader in [("train", train_loader), ("val", val_loader), ("test", test_loader)]:
    X_batch, y_batch = next(iter(loader))
    print(f"{split_name}: X batch shape={X_batch.shape}, y batch shape={y_batch.shape}")
    assert X_batch.dim() == 3
    assert X_batch.shape[1] == actual_seq_len
    assert X_batch.shape[2] == actual_feature_count
```

```python
# ============================================================================
# Dynamic Model Architectures
# ============================================================================

class DenseModel(nn.Module):
    def __init__(self, input_size, hidden_size=128, dropout=0.3):
        super().__init__()
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(input_size, hidden_size)
        self.dropout1 = nn.Dropout(dropout)
        self.fc2 = nn.Linear(hidden_size, hidden_size // 2)
        self.dropout2 = nn.Dropout(dropout)
        self.fc3 = nn.Linear(hidden_size // 2, 1)
        self.relu = nn.ReLU()

    def forward(self, x):
        x = self.flatten(x)
        x = self.relu(self.fc1(x))
        x = self.dropout1(x)
        x = self.relu(self.fc2(x))
        x = self.dropout2(x)
        return self.fc3(x)

class LSTMModel(nn.Module):
    def __init__(self, input_size, hidden_size=64, num_layers=2, dropout=0.3):
        super().__init__()
        self.input_size = input_size
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            dropout=dropout if num_layers > 1 else 0.0,
            batch_first=True,
        )
        self.fc = nn.Sequential(
            nn.Linear(hidden_size, hidden_size // 2),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_size // 2, 1),
        )

    def forward(self, x):
        _, (hidden, _) = self.lstm(x)
        return self.fc(hidden[-1])

class TransformerModel(nn.Module):
    def __init__(self, input_size, hidden_size=128, num_heads=4, num_layers=2, dropout=0.3, max_seq_len=512):
        super().__init__()
        self.input_size = input_size
        if hidden_size % num_heads != 0:
            raise ValueError(f"hidden_size ({hidden_size}) must be divisible by num_heads ({num_heads})")

        self.input_projection = nn.Linear(input_size, hidden_size)
        self.position_embedding = nn.Parameter(torch.zeros(1, max_seq_len, hidden_size))
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=hidden_size,
            nhead=num_heads,
            dim_feedforward=hidden_size * 4,
            dropout=dropout,
            batch_first=True,
            activation="gelu",
        )
        self.encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        self.fc = nn.Sequential(
            nn.LayerNorm(hidden_size),
            nn.Linear(hidden_size, hidden_size // 2),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_size // 2, 1),
        )

    def forward(self, x):
        seq_len = x.shape[1]
        if seq_len > self.position_embedding.shape[1]:
            raise ValueError(f"Sequence length {seq_len} exceeds max_seq_len {self.position_embedding.shape[1]}")
        x = self.input_projection(x)
        x = x + self.position_embedding[:, :seq_len, :]
        x = self.encoder(x)
        return self.fc(x[:, -1, :])

print(f"Dense input_size from X.shape: {actual_input_size}")
print(f"Sequential model input_size from X.shape[2]: {actual_feature_count}")
```

```python
# ============================================================================
# Runtime Shape Validation and Training Utilities
# ============================================================================

def validate_model_input_shape(model, X_batch, model_name):
    assert X_batch.dim() == 3

    if hasattr(model, "fc1"):
        expected_size = model.fc1.in_features
        actual_size = X_batch.view(X_batch.size(0), -1).shape[1]

        if actual_size != expected_size:
            raise ValueError(f"Shape mismatch: expected {expected_size}, got {actual_size}")

    if hasattr(model, "input_size"):
        expected_features = model.input_size
        actual_features = X_batch.shape[2]
        if actual_features != expected_features:
            raise ValueError(
                f"{model_name} feature mismatch: expected {expected_features}, got {actual_features}"
            )

def log_first_batch_shape(model, dataloader, model_name):
    X_batch, _ = next(iter(dataloader))
    print(f"\n{model_name} runtime shape validation")
    print("X batch shape:", X_batch.shape)
    print("Flattened size:", X_batch.view(X_batch.size(0), -1).shape)
    assert X_batch.dim() == 3
    validate_model_input_shape(model, X_batch, model_name)

def forward_checked(model, X_batch, model_name):
    validate_model_input_shape(model, X_batch, model_name)
    return model(X_batch)

def train_model(model, model_name, train_loader, val_loader, epochs=EPOCHS, learning_rate=LEARNING_RATE, patience=PATIENCE):
    model = model.to(device)
    log_first_batch_shape(model, train_loader, model_name)

    optimizer = optim.AdamW(model.parameters(), lr=learning_rate, weight_decay=1e-5)
    loss_fn = nn.MSELoss()
    scaler = torch.cuda.amp.GradScaler(enabled=USE_AMP)

    best_val_loss = float("inf")
    best_state = None
    patience_counter = 0
    train_losses = []
    val_losses = []

    for epoch in tqdm(range(1, epochs + 1), desc=f"Training {model_name}"):
        model.train()
        train_loss_total = 0.0
        train_samples = 0

        for X_batch, y_batch in train_loader:
            assert X_batch.dim() == 3
            X_batch = X_batch.to(device, non_blocking=PIN_MEMORY)
            y_batch = y_batch.to(device, non_blocking=PIN_MEMORY)

            optimizer.zero_grad(set_to_none=True)
            with autocast_context():
                predictions = forward_checked(model, X_batch, model_name)
                loss = loss_fn(predictions, y_batch)

            if USE_AMP:
                scaler.scale(loss).backward()
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                scaler.step(optimizer)
                scaler.update()
            else:
                loss.backward()
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                optimizer.step()

            batch_size = X_batch.size(0)
            train_loss_total += loss.item() * batch_size
            train_samples += batch_size

        train_loss = train_loss_total / max(train_samples, 1)
        val_loss = evaluate_loss(model, model_name, val_loader, loss_fn)
        train_losses.append(train_loss)
        val_losses.append(val_loss)

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_state = {key: value.detach().cpu().clone() for key, value in model.state_dict().items()}
            patience_counter = 0
        else:
            patience_counter += 1

        if epoch == 1 or epoch % 5 == 0:
            print(f"{model_name} epoch {epoch:03d}: train_loss={train_loss:.6f}, val_loss={val_loss:.6f}")

        if patience_counter >= patience:
            print(f"{model_name}: early stopping at epoch {epoch}")
            break

    if best_state is not None:
        model.load_state_dict(best_state)

    return model, train_losses, val_losses

def evaluate_loss(model, model_name, dataloader, loss_fn):
    model.eval()
    loss_total = 0.0
    samples = 0
    with torch.no_grad():
        for X_batch, y_batch in dataloader:
            assert X_batch.dim() == 3
            X_batch = X_batch.to(device, non_blocking=PIN_MEMORY)
            y_batch = y_batch.to(device, non_blocking=PIN_MEMORY)
            with autocast_context():
                predictions = forward_checked(model, X_batch, model_name)
                loss = loss_fn(predictions, y_batch)
            batch_size = X_batch.size(0)
            loss_total += loss.item() * batch_size
            samples += batch_size
    return loss_total / max(samples, 1)
```

```python
# ============================================================================
# Train Dense, LSTM, and Transformer Models
# ============================================================================

models = {
    "Dense": DenseModel(input_size=actual_input_size, hidden_size=128, dropout=0.3),
    "LSTM": LSTMModel(input_size=actual_feature_count, hidden_size=64, num_layers=2, dropout=0.3),
    "Transformer": TransformerModel(
        input_size=actual_feature_count,
        hidden_size=128,
        num_heads=4,
        num_layers=2,
        dropout=0.3,
        max_seq_len=max(SEQ_LEN, actual_seq_len),
    ),
}

trained_models = {}
training_history = {}
model_failures = {}

for model_name, model in models.items():
    try:
        trained_model, train_losses, val_losses = train_model(model, model_name, train_loader, val_loader)
        trained_models[model_name] = trained_model
        training_history[model_name] = {"train": train_losses, "val": val_losses}
    except Exception as exc:
        model_failures[model_name] = str(exc)
        print(f"Skipping model {model_name}: {exc}")

if not trained_models:
    raise RuntimeError(f"No models trained successfully: {model_failures}")

print(f"Trained models: {list(trained_models)}")
print(f"Model failures: {model_failures}")
```

```python

# ============================================================================
# Prediction, Ensemble, and Metrics
# ============================================================================

def predict_model(model, model_name, dataloader):
    model.eval()
    predictions = []
    actuals = []
    with torch.no_grad():
        for X_batch, y_batch in dataloader:
            assert X_batch.dim() == 3
            X_batch = X_batch.to(device, non_blocking=PIN_MEMORY)
            with autocast_context():
                batch_predictions = forward_checked(model, X_batch, model_name)
            predictions.append(batch_predictions.detach().cpu().numpy().reshape(-1))
            actuals.append(y_batch.detach().cpu().numpy().reshape(-1))
    return np.concatenate(predictions), np.concatenate(actuals)

def predict_single_model(model, model_name, X_input):
    model.eval()
    X_tensor = torch.as_tensor(X_input, dtype=torch.float32).to(device)
    with torch.no_grad():
        with autocast_context():
            prediction = forward_checked(model, X_tensor, model_name)
    return prediction.detach().cpu().numpy().reshape(-1).astype(np.float32)

def normalize_ensemble_weights(model_names, weights=None):
    model_names = list(model_names)
    if weights is None:
        return {name: 1.0 / len(model_names) for name in model_names}
    available = {name: max(0.0, float(weights.get(name, 0.0))) for name in model_names}
    total_weight = sum(available.values())
    if total_weight <= 0:
        return {name: 1.0 / len(model_names) for name in model_names}
    return {name: weight / total_weight for name, weight in available.items()}

def ensemble_predict(trained_models, dataloader, weights=None):
    normalized_weights = normalize_ensemble_weights(trained_models.keys(), weights)
    model_predictions = {}
    actuals = None
    for model_name, model in trained_models.items():
        preds, y_true = predict_model(model, model_name, dataloader)
        model_predictions[model_name] = preds
        if actuals is None:
            actuals = y_true
        elif len(actuals) != len(y_true):
            raise ValueError(f"Prediction alignment mismatch for {model_name}")

    ensemble = np.zeros_like(actuals, dtype=np.float32)
    for model_name in trained_models:
        preds = model_predictions[model_name]
        if len(preds) != len(ensemble):
            raise ValueError(f"Prediction length mismatch for {model_name}")
        ensemble += normalized_weights[model_name] * preds

    return ensemble, actuals, model_predictions, normalized_weights

def ensemble_single_predictions(model_predictions, weights=None):
    normalized_weights = normalize_ensemble_weights(model_predictions.keys(), weights)
    ensemble_value = 0.0
    for model_name in model_predictions:
        prediction = np.asarray(model_predictions[model_name], dtype=np.float32).reshape(-1)
        if len(prediction) != 1:
            raise ValueError(f"Expected one prediction for {model_name}, got {len(prediction)}")
        ensemble_value += normalized_weights[model_name] * float(prediction[0])
    return float(ensemble_value), normalized_weights

def metric_value(value):
    if value is None or not np.isfinite(value):
        return None
    return float(value)

def calculate_metrics(y_true, y_pred, label, predictor_count=actual_input_size):
    y_true = np.asarray(y_true, dtype=np.float64).reshape(-1)
    y_pred = np.asarray(y_pred, dtype=np.float64).reshape(-1)
    valid = np.isfinite(y_true) & np.isfinite(y_pred)
    y_true = y_true[valid]
    y_pred = y_pred[valid]
    n = len(y_true)
    if n == 0:
        print(f"{label}: no valid rows")
        return {"n": 0, "MSE": None, "RMSE": None, "MAE": None, "R2": None, "Adjusted_R2": None, "MAPE": None}

    mse = mean_squared_error(y_true, y_pred)
    rmse = math.sqrt(mse)
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred) if n > 1 and len(np.unique(y_true)) > 1 else np.nan
    adjusted_r2 = np.nan
    if n > predictor_count + 1 and np.isfinite(r2):
        adjusted_r2 = 1 - (1 - r2) * (n - 1) / (n - predictor_count - 1)
    nonzero = y_true != 0
    mape = np.mean(np.abs((y_true[nonzero] - y_pred[nonzero]) / y_true[nonzero])) * 100 if nonzero.any() else np.nan

    print(f"{label}: RMSE={rmse:.6f}, MAE={mae:.6f}, R2={r2:.6f}")
    return {
        "n": int(n),
        "MSE": metric_value(mse),
        "RMSE": metric_value(rmse),
        "MAE": metric_value(mae),
        "R2": metric_value(r2),
        "Adjusted_R2": metric_value(adjusted_r2),
        "MAPE": metric_value(mape),
    }

ensemble_weights = {"Dense": 0.2, "LSTM": 0.3, "Transformer": 0.5}

train_pred, train_actual, train_model_preds, used_weights = ensemble_predict(trained_models, train_loader, ensemble_weights)
val_pred, val_actual, val_model_preds, _ = ensemble_predict(trained_models, val_loader, ensemble_weights)

metrics = {
    "Train": calculate_metrics(train_actual, train_pred, "Train scaled metrics"),
    "Validation": calculate_metrics(val_actual, val_pred, "Validation scaled metrics"),
}

print(f"Ensemble weights used: {used_weights}")
```

```python

# ============================================================================
# Leakage-Guarded Walk-Forward Evaluation
# ============================================================================

def inverse_scale_by_symbol(values, symbols, scalers):
    inverse = np.empty_like(values, dtype=np.float32)
    for symbol in np.unique(symbols):
        mask = symbols == symbol
        scaler = scalers.get(symbol)
        if scaler is None:
            inverse[mask] = values[mask]
        else:
            inverse[mask] = scaler.inverse_transform(values[mask].reshape(-1, 1)).reshape(-1)
    return inverse

def get_row_date(df, date_col, row_idx):
    if date_col and date_col in df.columns:
        return df.iloc[row_idx][date_col]
    return row_idx

def prepare_indicator_input(df):
    rename_map = {
        "Open": "open",
        "High": "high",
        "Low": "low",
        "Close": "close",
        "Volume": "volume",
    }
    return df.rename(columns={old: new for old, new in rename_map.items() if old in df.columns})

def recompute_history_features(history_df, date_col, feature_columns):
    indicator_input = prepare_indicator_input(history_df.copy())
    indicator_date_col = date_col
    with contextlib.redirect_stdout(sys.stderr):
        engineered = compute_indicators(indicator_input, indicator_date_col)
    engineered.columns = [canonical_column_name(col) for col in engineered.columns]
    engineered = engineered.loc[:, ~pd.Index(engineered.columns).duplicated()].copy()
    if not {"Open", "High", "Low", "Close", "Volume"}.issubset(set(engineered.columns)):
        raise ValueError("OHLCV columns missing after indicator recomputation")
    return clean_aligned_stock_frame(engineered, feature_columns, TARGET_COL)

def build_walk_forward_input(history_df, current_open, target_idx, feature_columns, feature_scaler, seq_len, date_col):
    if len(history_df) < seq_len:
        return None
    if len(history_df) > target_idx:
        raise ValueError(f"History includes target/future rows: len(history)={len(history_df)}, target_idx={target_idx}")

    engineered_history = recompute_history_features(history_df, date_col, feature_columns)
    if len(engineered_history) < seq_len:
        return None

    history_indices = np.arange(target_idx - seq_len, target_idx, dtype=np.int64)
    validate_no_leakage_state(history_indices, target_idx, feature_columns)

    history_features = engineered_history[feature_columns].iloc[-seq_len:]
    scaled_history = feature_scaler.transform(history_features).astype(np.float32)
    X_input = scaled_history.reshape(1, seq_len, len(feature_columns)).copy()

    if USE_INTRADAY_OPEN:
        open_idx = feature_columns.index("Open")
        before_seq = X_input[0].copy()
        current_features = pd.DataFrame(
            np.zeros((1, len(feature_columns)), dtype=np.float32),
            columns=list(feature_columns),
        )
        current_features.loc[:, "Open"] = float(current_open)
        open_t_scaled = feature_scaler.transform(current_features).astype(np.float32)[0, open_idx]
        X_input[0, -1, open_idx] = open_t_scaled
        validate_intraday_open_injection(before_seq, X_input[0], open_idx)

    return X_input.astype(np.float32)

def assert_sentinel_no_future_close_dependency(full_df, target_idx, feature_columns, feature_scaler, seq_len, date_col):
    if not RUN_SENTINEL_LEAKAGE_CHECK:
        return
    current_open = full_df.iloc[target_idx]["Open"]
    original_history = full_df.iloc[:target_idx].copy()
    sentinel_df = full_df.copy()
    sentinel_df.loc[sentinel_df.index[target_idx:], TARGET_COL] = SENTINEL_CLOSE_VALUE
    sentinel_history = sentinel_df.iloc[:target_idx].copy()

    original_X = build_walk_forward_input(original_history, current_open, target_idx, feature_columns, feature_scaler, seq_len, date_col)
    sentinel_X = build_walk_forward_input(sentinel_history, current_open, target_idx, feature_columns, feature_scaler, seq_len, date_col)
    if original_X is None or sentinel_X is None:
        return
    if not np.allclose(original_X, sentinel_X, rtol=0.0, atol=1e-7, equal_nan=True):
        raise ValueError(f"Sentinel leakage check failed at target_idx={target_idx}")

def optional_online_update(trained_models, X_input, actual_scaled):
    if not USE_ONLINE_FINE_TUNING:
        return
    loss_fn = nn.MSELoss()
    X_tensor = torch.as_tensor(X_input, dtype=torch.float32).to(device)
    y_tensor = torch.as_tensor([[float(actual_scaled)]], dtype=torch.float32).to(device)
    for model_name, model in trained_models.items():
        model.train()
        optimizer = optim.AdamW(model.parameters(), lr=ONLINE_LEARNING_RATE, weight_decay=0.0)
        optimizer.zero_grad(set_to_none=True)
        prediction = forward_checked(model, X_tensor, model_name)
        loss = loss_fn(prediction, y_tensor)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        model.eval()

def run_walk_forward_simulation(trained_models, artifacts, weights):
    rows = []
    for symbol, artifact in tqdm(artifacts.items(), desc="Walk-forward evaluation"):
        full_df = artifact["source_frame"].reset_index(drop=True).copy()
        date_col = artifact["date_col"]
        feature_scaler = artifact["feature_scaler"]
        target_scaler = artifact["target_scaler"]
        test_indices = np.asarray(artifact["test_target_indices"], dtype=np.int64)
        feature_snapshot = snapshot_scaler(feature_scaler)
        target_snapshot = snapshot_scaler(target_scaler)
        sentinel_checked = False

        if len(test_indices) == 0:
            continue

        available_history = full_df.iloc[: int(test_indices[0])].copy()

        for target_idx in test_indices:
            target_idx = int(target_idx)
            if len(available_history) > target_idx:
                raise ValueError(f"{symbol}: available history leaked into target row {target_idx}")
            if len(available_history) < target_idx:
                raise ValueError(f"{symbol}: walk-forward history gap before target row {target_idx}")

            current_open = pd.to_numeric(pd.Series([full_df.iloc[target_idx]["Open"]]), errors="coerce").iloc[0]
            actual_close = pd.to_numeric(pd.Series([full_df.iloc[target_idx][TARGET_COL]]), errors="coerce").iloc[0]
            if not np.isfinite(current_open) or not np.isfinite(actual_close):
                available_history = pd.concat([available_history, full_df.iloc[[target_idx]]], ignore_index=True)
                continue

            if len(available_history) < SEQ_LEN:
                available_history = pd.concat([available_history, full_df.iloc[[target_idx]]], ignore_index=True)
                continue

            if RUN_SENTINEL_LEAKAGE_CHECK and not sentinel_checked:
                assert_sentinel_no_future_close_dependency(full_df, target_idx, common_features, feature_scaler, SEQ_LEN, date_col)
                sentinel_checked = True

            X_input = build_walk_forward_input(
                available_history,
                current_open,
                target_idx,
                common_features,
                feature_scaler,
                SEQ_LEN,
                date_col,
            )
            if X_input is None:
                available_history = pd.concat([available_history, full_df.iloc[[target_idx]]], ignore_index=True)
                continue

            model_predictions = {}
            for model_name, model in trained_models.items():
                model_predictions[model_name] = predict_single_model(model, model_name, X_input)

            ensemble_scaled, row_weights = ensemble_single_predictions(model_predictions, weights)
            actual_scaled = target_scaler.transform(pd.DataFrame({TARGET_COL: [actual_close]})).astype(np.float32).reshape(-1)[0]
            predicted_price = target_scaler.inverse_transform(np.asarray([[ensemble_scaled]], dtype=np.float32)).reshape(-1)[0]

            row = {
                "Symbol": symbol,
                "Target_Index": target_idx,
                "Date": get_row_date(full_df, date_col, target_idx),
                "Actual_Scaled": float(actual_scaled),
                "Predicted_Scaled": float(ensemble_scaled),
                "Actual_Price": float(actual_close),
                "Predicted_Price": float(predicted_price),
                "Absolute_Error_Price": float(abs(actual_close - predicted_price)),
            }
            for model_name, prediction in model_predictions.items():
                row[f"{model_name}_Scaled"] = float(np.asarray(prediction).reshape(-1)[0])
            rows.append(row)

            optional_online_update(trained_models, X_input, actual_scaled)
            available_history = pd.concat([available_history, full_df.iloc[[target_idx]]], ignore_index=True)

        assert_scaler_unchanged(f"{symbol} feature_scaler", feature_scaler, feature_snapshot)
        assert_scaler_unchanged(f"{symbol} target_scaler", target_scaler, target_snapshot)
        validate_scaler_fit_scope(feature_scaler, artifact["train_end_idx"], f"{symbol} feature_scaler")
        validate_scaler_fit_scope(target_scaler, artifact["train_end_idx"], f"{symbol} target_scaler")

    return pd.DataFrame(rows)

results_df = run_walk_forward_simulation(trained_models, stock_walk_forward_artifacts, used_weights)
if results_df.empty:
    raise RuntimeError("Walk-forward evaluation produced no predictions")

test_actual = results_df["Actual_Scaled"].to_numpy(dtype=np.float32)
test_pred = results_df["Predicted_Scaled"].to_numpy(dtype=np.float32)
test_symbols = results_df["Symbol"].to_numpy(dtype=object)
test_actual_price = results_df["Actual_Price"].to_numpy(dtype=np.float32)
test_pred_price = results_df["Predicted_Price"].to_numpy(dtype=np.float32)

metrics["Test"] = calculate_metrics(test_actual, test_pred, "Walk-forward test scaled metrics")
price_metrics = calculate_metrics(test_actual_price, test_pred_price, "Walk-forward test price metrics")

results_df.head()
```
